```python
import pandas as pd

# Load the CSV file
file_path = 'NASI.csv'  # Replace with your actual file path if different
df = pd.read_csv(file_path)

# Remove rows where 'orig_reason' is empty or NaN
filtered_df = df.dropna(subset=['orig_reason'])

# Save the filtered data back to a CSV file
output_path = 'NASI_filtered.csv'  # Save to a new file
filtered_df.to_csv(output_path, index=False)

output_path
```

```python
# Mapping of text to their respective replacements
replacement_map = {
    "Good state": "C1",
    "Jobs and careers related stressed": "C2",
    "Medication related stressed": "C3",
    "Relationship related stressed": "C4",
    "Alienation related stressed": "C5",
    "Bias or abuse related stressed": "C6"
}

# Replace values in the 'category' column based on the mapping
df['category'] = df['category'].replace(replacement_map)

# Save the updated DataFrame to a new file
updated_file_path = 'NASI_updated.csv'
df.to_csv(updated_file_path, index=False)

updated_file_path

```
